```python
#FISHER MATRIX for wowaCDM model
import numpy as np
from scipy.integrate import quad
#costant variables
Or=0
c=299792
#PARAMETERS
#wo=-0.957 #plank 2018
#wa=-0.29
#(rs*H0)=10000
#Om=0.321
wo= -0.55 #mine from cobaya
wa= -1.6
rs = 140
H0 = 70
Om = 0.34
#define the functions
def fde(z,wo,wa): #dark energy density
    return (1/(1+z))**(-3*(1+wo+wa))*np.exp(-3*wa*(z/(1+z)))
def H(z,wo,wa,Om): 
    return np.sqrt(Om*(1+z)**3+Or*(1+z)**4+(1-Om-Or)*fde(z,wo,wa))
def Hi(z,wo,wa,Om):
    return 1/np.sqrt(Om*(1+z)**3+Or*(1+z)**4+(1-Om-Or)*fde(z,wo,wa))
def DH(z,wo,wa,Om, rs, H0): #line of sight distance
    return c/H(z,wo,wa,Om)/(rs*H0)
def DM(z,wo,wa,Om, rs, H0): #transverse distance
    I,_=quad(Hi, 0, z, args=(wo,wa,Om))
    return c*I/(rs*H0)
def DV(z,wo,wa,Om, rs, H0): #angle averaged distance
    return (z*DM(z,wo,wa,Om, rs, H0)**2*DH(z,wo,wa,Om, rs, H0))**(1/3)
# Calculate D_M(z) / r_d based on Eq. 2.9
def DL(z,wo,wa,Om, H0):
    I,_=quad(Hi, 0, z, args=(wo,wa,Om))
    return 5 * np.log10( c * I * (1+z) /H0 )+25
    
```

```python
# Fourth-order derivative with respect to wo for DH
from scipy.integrate import quad
import numpy as np

def DH_dwo(z, wo, wa, Om, rs, H0):
    h = wo / 100 
    return (-DH(z, wo + 2*h, wa, Om, rs, H0) + 8*DH(z, wo + h, wa, Om, rs, H0) 
            - 8*DH(z, wo - h, wa, Om, rs, H0) + DH(z, wo - 2*h, wa, Om, rs, H0)) / (12 * h)

def DH_dwa(z, wo, wa, Om, rs, H0):
    h = wa / 100
    return (-DH(z, wo, wa + 2*h, Om, rs, H0) + 8*DH(z, wo, wa + h, Om, rs, H0) 
            - 8*DH(z, wo, wa - h, Om, rs, H0) + DH(z, wo, wa - 2*h, Om, rs, H0)) / (12 * h)

def DH_dOm(z, wo, wa, Om, rs, H0):
    h = Om / 100 
    return (-DH(z, wo, wa, Om + 2*h, rs, H0) + 8*DH(z, wo, wa, Om + h, rs, H0) 
            - 8*DH(z, wo, wa, Om - h, rs, H0) + DH(z, wo, wa, Om - 2*h, rs, H0)) / (12 * h)

def DH_d_rs(z, wo, wa, Om, rs, H0):
    h = rs / 100
    return (-DH(z, wo, wa, Om, rs + 2*h, H0) + 8*DH(z, wo, wa, Om, rs + h, H0) 
            - 8*DH(z, wo, wa, Om, rs - h, H0) + DH(z, wo, wa, Om, rs - 2*h, H0)) / (12 * h)

def DH_d_H0(z, wo, wa, Om, rs, H0):
    h = H0 / 100
    return (-DH(z, wo, wa, Om, rs, H0 + 2*h) + 8*DH(z, wo, wa, Om, rs, H0 + h) 
            - 8*DH(z, wo, wa, Om, rs, H0 - h) + DH(z, wo, wa, Om, rs, H0 - 2*h)) / (12 * h)

def DM_dwo(z, wo, wa, Om, rs, H0):
    h = wo / 100 
    return (-DM(z, wo + 2*h, wa, Om, rs, H0) + 8*DM(z, wo + h, wa, Om, rs, H0) 
            - 8*DM(z, wo - h, wa, Om, rs, H0) + DM(z, wo - 2*h, wa, Om, rs, H0)) / (12 * h)

def DM_dwa(z, wo, wa, Om, rs, H0):
    h = wa / 100 
    return (-DM(z, wo, wa + 2*h, Om, rs, H0) + 8*DM(z, wo, wa + h, Om, rs, H0) 
            - 8*DM(z, wo, wa - h, Om, rs, H0) + DM(z, wo, wa - 2*h, Om, rs, H0)) / (12 * h)

def DM_dOm(z, wo, wa, Om, rs, H0):
    h = Om / 100 
    return (-DM(z, wo, wa, Om + 2*h, rs, H0) + 8*DM(z, wo, wa, Om + h, rs, H0) 
            - 8*DM(z, wo, wa, Om - h, rs, H0) + DM(z, wo, wa, Om - 2*h, rs, H0)) / (12 * h)
    
def DM_d_rs(z, wo, wa, Om, rs, H0):
    h = rs / 100
    return (-DM(z, wo, wa, Om, rs + 2*h, H0) + 8*DM(z, wo, wa, Om, rs + h, H0) 
            - 8*DM(z, wo, wa, Om, rs - h, H0) + DM(z, wo, wa, Om, rs - 2*h, H0)) / (12 * h)

def DM_d_H0(z, wo, wa, Om, rs, H0):
    h = H0 / 100
    return (-DM(z, wo, wa, Om, rs, H0 + 2*h) + 8*DM(z, wo, wa, Om, rs, H0 + h) 
            - 8*DM(z, wo, wa, Om, rs, H0 - h) + DM(z, wo, wa, Om, rs, H0 - 2*h)) / (12 * h)


def DV_dwo(z, wo, wa, Om, rs, H0):
    h = wo / 100 
    return (-DV(z, wo + 2*h, wa, Om, rs, H0) + 8*DV(z, wo + h, wa, Om, rs, H0) 
            - 8*DV(z, wo - h, wa, Om, rs, H0) + DV(z, wo - 2*h, wa, Om, rs, H0)) / (12 * h)

def DV_dwa(z, wo, wa, Om, rs, H0):
    h = wa / 100 
    return (-DV(z, wo, wa + 2*h, Om, rs, H0) + 8*DV(z, wo, wa + h, Om, rs, H0) 
            - 8*DV(z, wo, wa - h, Om, rs, H0) + DV(z, wo, wa - 2*h, Om, rs, H0)) / (12 * h)

def DV_dOm(z, wo, wa, Om, rs, H0):
    h = Om / 100 
    return (-DV(z, wo, wa, Om + 2*h, rs, H0) + 8*DV(z, wo, wa, Om + h, rs, H0) 
            - 8*DV(z, wo, wa, Om - h, rs, H0) + DV(z, wo, wa, Om - 2*h, rs, H0)) / (12 * h)

def DV_d_rs(z, wo, wa, Om, rs, H0):
    h = rs / 100
    return (-DV(z, wo, wa, Om, rs + 2*h, H0) + 8*DV(z, wo, wa, Om, rs + h, H0) 
            - 8*DV(z, wo, wa, Om, rs - h, H0) + DV(z, wo, wa, Om, rs - 2*h, H0)) / (12 * h)

def DV_d_H0(z, wo, wa, Om, rs, H0):
    h = H0 / 100
    return (-DV(z, wo, wa, Om, rs, H0 + 2*h) + 8*DV(z, wo, wa, Om, rs, H0 + h) 
            - 8*DV(z, wo, wa, Om, rs, H0 - h) + DV(z, wo, wa, Om, rs, H0 - 2*h)) / (12 * h)


def DL_dwo(z, wo, wa, Om, H0):
    h = wo / 100 
    return (-DL(z, wo + 2*h, wa, Om, H0) + 8*DL(z, wo + h, wa, Om, H0) 
            - 8*DL(z, wo - h, wa, Om, H0) + DL(z, wo - 2*h, wa, Om, H0)) / (12 * h)

def DL_dwa(z, wo, wa, Om, H0):
    h = wa / 100 
    return (-DL(z, wo, wa + 2*h, Om, H0) + 8*DL(z, wo, wa + h, Om, H0) 
            - 8*DL(z, wo, wa - h, Om, H0) + DL(z, wo, wa - 2*h, Om, H0)) / (12 * h)

def DL_dOm(z, wo, wa, Om, H0):
    h = Om / 100 
    return (-DL(z, wo, wa, Om + 2*h, H0) + 8*DL(z, wo, wa, Om + h, H0) 
            - 8*DL(z, wo, wa, Om - h, H0) + DL(z, wo, wa, Om - 2*h, H0)) / (12 * h)

def DL_d_H0(z, wo, wa, Om, H0):
    h = H0 / 100 
    return (-DL(z, wo, wa, Om, H0 + 2*h) + 8*DL(z, wo, wa, Om, H0 + h) 
            - 8*DL(z, wo, wa, Om, H0 - h) + DL(z, wo, wa, Om, H0 - 2*h)) / (12 * h)
```

```python
zvals = (0.295,1.491,0.510, 0.706, 0.930, 1.317, 2.330)
npar = 5
F = np.zeros([npar, npar])
for zcount in range(2, 7):#add DM,DH
    z = zvals[zcount]
    for i in range(npar): #cicle through matrix element
        dfdpi = [0, 0]
        if i == 0: 
            dfdpi[0] = DM_dwo(z, wo, wa, Om, rs, H0)
            dfdpi[1] = DH_dwo(z, wo, wa, Om, rs, H0)
        elif i == 1:
            dfdpi[0] = DM_dwa(z, wo, wa, Om, rs, H0)
            dfdpi[1] = DH_dwa(z, wo, wa, Om, rs, H0)
        elif i == 2: 
            dfdpi[0] = DM_dOm(z, wo, wa, Om, rs, H0)
            dfdpi[1] = DH_dOm(z, wo, wa, Om, rs, H0)
        elif i == 3:
            dfdpi[0] = DM_d_H0(z, wo, wa, Om, rs, H0)
            dfdpi[1] = DH_d_H0(z, wo, wa, Om, rs, H0)
        elif i == 4:
            dfdpi[0] = DM_d_rs(z, wo, wa, Om, rs, H0)
            dfdpi[1] = DH_d_rs(z, wo, wa, Om, rs, H0)

        for j in range(npar):
            dfdpj = [0, 0]
            if j == 0: 
                dfdpj[0] = DM_dwo(z, wo, wa, Om, rs, H0)
                dfdpj[1] = DH_dwo(z, wo, wa, Om, rs, H0)
            elif j == 1:
                dfdpj[0] = DM_dwa(z, wo, wa, Om, rs, H0)
                dfdpj[1] = DH_dwa(z, wo, wa, Om, rs, H0)
            elif j == 2: 
                dfdpj[0] = DM_dOm(z, wo, wa, Om, rs, H0)
                dfdpj[1] = DH_dOm(z, wo, wa, Om, rs, H0)
            elif j == 3:
                dfdpj[0] = DM_d_H0(z, wo, wa, Om, rs, H0)
                dfdpj[1] = DH_d_H0(z, wo, wa, Om, rs, H0)
            elif j == 4:
                dfdpj[0] = DM_d_rs(z, wo, wa, Om, rs, H0)
                dfdpj[1] = DH_d_rs(z, wo, wa, Om, rs, H0)
  
            for i2 in range(2): #sum DH and DM for matrix element with the respective variances and covariances
                for j2 in range(2):
                    F[i, j] += dfdpi[i2] * dfdpj[j2] * Si[i2 + 2 *(zcount-1), j2 + 2 * (zcount-1)]
                    
for zcount in range(2):  #add DV
    z = zvals[zcount]
    for i in range(npar): #cicle through matrix element
        if i == 0: 
            dfdpi = DV_dwo(z, wo, wa, Om, rs, H0)
        elif i == 1:
            dfdpi = DV_dwa(z, wo, wa, Om, rs, H0)
        elif i == 2: 
            dfdpi = DV_dOm(z, wo, wa, Om, rs, H0)
        elif i == 3:
            dfdpi = DV_d_H0(z, wo, wa, Om, rs, H0)
        elif i == 4:
            dfdpi = DV_d_rs(z, wo, wa, Om, rs, H0)
        for j in range(npar):
            if j == 0: 
                dfdpj = DV_dwo(z, wo, wa, Om, rs, H0)
            elif j == 1:
                dfdpj = DV_dwa(z, wo, wa, Om, rs, H0)
            elif j == 2: 
                dfdpj = DV_dOm(z, wo, wa, Om, rs, H0)
            elif j == 3:
                dfdpj = DV_d_H0(z, wo, wa, Om, rs, H0)
            elif j == 4:
                dfdpj = DV_d_rs(z, wo, wa, Om, rs, H0)
            F[i,j] += Si[zcount,zcount]*dfdpi*dfdpj


print(F)
#print(np.linalg.inv(F))  # invert the matrix
#print(np.sqrt(np.abs(np.linalg.inv(F))))  # square it
```
